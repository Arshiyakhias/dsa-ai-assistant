Define a url and send a GET request

```python
url = "https://worldpopulationreview.com/country-rankings/largest-countries-in-the-world"

print(url)
```

```python
response=requests.get(url)
print(response)
```

Checking the status code

```python
print(response.status_code)
```

| Code  | Meaning                 |
| ----- | ----------------------- |
| `200` | Request successful      |
| `300` | Redirection             |
| `400` | Client-side error       |
| `404` | Page/resource not found |
| `500` | Server-side error       |


Demonstrating 404 error

```python
bad_url = "https://worldpopulationreview.com/this-page-does-not-exist"

response_404 = requests.get(bad_url)
```

```python
print(response_404.status_code)
```
