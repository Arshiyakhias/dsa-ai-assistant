```python
from bs4 import BeautifulSoup
```

```python
import requests

url = "https://worldpopulationreview.com/country-rankings/largest-countries-in-the-world"

response = requests.get(url)

print(response.status_code)
```

Create the Beautiful Soup object

A parser is a program that takes raw data and converts it into a structured form that a program can understand and work with.

response.content is just the raw HTML, and BeautifulSoup needs to understand the structure before you can search through it there fore it needs to go through a parser.
| Parser        |
| ------------- |
| `html.parser` |
| `lxml`        |
| `html5lib`    |
| `xml`         |


```python
soup = BeautifulSoup(response.content, "html.parser")
```

Why do we need teh Beautiful Soup

```python
print(response.text[:1000])
```
