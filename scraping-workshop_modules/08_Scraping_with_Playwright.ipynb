Scraping with Playwright

`requests` never runs JavaScript, so it can't see content that a page builds after loading. Playwright fixes that by driving a real browser - JavaScript actually executes, and we read the page once it's finished rendering.

Launching a browser and visiting the JS-rendered page

```python
from playwright.sync_api import sync_playwright

URL = "http://quotes.toscrape.com/js/"
```

```python
with sync_playwright() as p:
    browser = p.chromium.launch(headless=True)
    page = browser.new_page()

    page.goto(URL, wait_until="networkidle")  # wait for the JS to finish

    quote_elements = page.locator(".quote")
    count = quote_elements.count()

    print(f"Found {count} quotes on the page.")
```

`wait_until="networkidle"` tells Playwright to wait until the page stops making network requests - a reasonable stand-in for "the JavaScript has probably finished running."

Extracting the quotes, same as before

```python
with sync_playwright() as p:
    browser = p.chromium.launch(headless=True)
    page = browser.new_page()
    page.goto(URL, wait_until="networkidle")

    quote_elements = page.locator(".quote")
    count = quote_elements.count()

    for i in range(min(3, count)):  # just the first 3
        el = quote_elements.nth(i)
        text = el.locator(".text").inner_text()
        author = el.locator(".author").inner_text()
        tags = el.locator(".tags .tag").all_inner_texts()
        print(f"{text}\n  - {author}  {tags}\n")

    browser.close()
```

Same URL, same task - the only thing that changed is giving JavaScript time to run before we looked at the page. Always call `browser.close()` when you're done, or the process stays running in the background.
