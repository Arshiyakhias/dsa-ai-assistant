Multi-Page Scraping, Saved to JSON

Same static-page technique from the first notebook, but looped across every page of quotes.toscrape.com, with the results saved to a file instead of just printed.

```python
import json
import requests
from bs4 import BeautifulSoup

BASE_URL = "http://quotes.toscrape.com/page/{}/"
```

The site doesn't tell us up front how many pages exist, so we keep requesting the next page number until one comes back with no quotes on it - that's our signal to stop.

```python
all_quotes = []
page_number = 1

while True:
    response = requests.get(BASE_URL.format(page_number))
    soup = BeautifulSoup(response.text, "html.parser")
    quote_elements = soup.select(".quote")

    if not quote_elements:
        break  # no more pages

    for q in quote_elements:
        all_quotes.append({
            "text": q.select_one(".text").get_text(strip=True),
            "author": q.select_one(".author").get_text(strip=True),
            "tags": [t.get_text(strip=True) for t in q.select(".tags .tag")],
        })

    print(f"Scraped page {page_number} ({len(quote_elements)} quotes)")
    page_number += 1
```

Save as JSON

```python
with open("quotes.json", "w", encoding="utf-8") as f:
    json.dump(all_quotes, f, indent=2, ensure_ascii=False)

print(f"Saved {len(all_quotes)} quotes total to quotes.json")
```

This is the same pagination pattern as the multi-chapter textbook scrape from notebook 06 - discover pages, loop until there's nothing left, save the result as structured data for whatever comes next.
