Turning Scraped Data into a Tiny Search Tool

No machine learning here - just a plain keyword filter over the JSON we saved in the last notebook. The point: raw scraped data can become something a person actually interacts with in a few lines of code.

```python
import json

with open("quotes.json", "r", encoding="utf-8") as f:
    quotes = json.load(f)

print(f"Loaded {len(quotes)} quotes.")
```

We check the keyword against the quote text, the author, and the tags - whichever one matches counts as a hit.

```python
keyword = input("Search quotes by keyword, author, or tag: ").strip().lower()

matches = [
    q for q in quotes
    if keyword in q["text"].lower()
    or keyword in q["author"].lower()
    or any(keyword in tag.lower() for tag in q["tags"])
]

print(f"Found {len(matches)} matching quote(s):")
```

```python
for q in matches:
    print(f'"{q["text"]}"\n  - {q["author"]}  {q["tags"]}\n')
```

That's the whole pipeline, start to finish: a GET request, a parser, a loop across pages, and now a search box - all built from the same handful of BeautifulSoup calls this course started with.
