Static Pages vs JavaScript-Rendered Pages

So far we've scraped pages where the server sends back the full HTML straight away. Not every site works like that - some rely on JavaScript to fill in the content *after* the page loads.

We'll use quotes.toscrape.com for this - it's a sandbox site built specifically for practicing scraping.

Part 1: A Static Page

```python
import requests
from bs4 import BeautifulSoup

URL = "http://quotes.toscrape.com/"
```

```python
response = requests.get(URL)
soup = BeautifulSoup(response.text, "html.parser")

print(response.status_code)
```

`.select()` grabs elements using a CSS selector - the same idea as `document.querySelectorAll()` in the browser console.

```python
quotes = soup.select(".quote")

print(f"Found {len(quotes)} quotes on the page.")
```

```python
for q in quotes[:3]:  # just the first 3
    text = q.select_one(".text").get_text(strip=True)
    author = q.select_one(".author").get_text(strip=True)
    tags = [t.get_text(strip=True) for t in q.select(".tags .tag")]
    print(f"{text}\n  - {author}  {tags}\n")
```

Part 2: The Same Technique on a JavaScript-Rendered Page

`quotes.toscrape.com/js/` looks identical in a browser, but the quotes are injected by a script *after* the page loads. `requests` only ever sees the HTML that arrives before that script runs.

```python
JS_URL = "http://quotes.toscrape.com/js/"

response = requests.get(JS_URL)
soup = BeautifulSoup(response.text, "html.parser")

quotes = soup.select(".quote")

print(f"Found {len(quotes)} quotes on the page.")
```

Zero quotes - and no error. The HTML `requests` sees is missing the JS-rendered content entirely.

Try it yourself: open `view-source:http://quotes.toscrape.com/js/` in a browser and search the page for the word "quote". It isn't there - which is exactly what our scraper just saw.
