```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://worldpopulationreview.com/country-rankings/largest-countries-in-the-world"

response = requests.get(url)

```

```python
soup = BeautifulSoup(response.content, "html.parser")
soup
```

```python
tables = soup.find_all("table")

print(len(tables))

for table in tables:
    print(table.get("class"))
```

```python
tabel_1 = tables[0]
```

```python
headers = tabel_1.find_all("th")
for header in headers:
    print(header.prettify())
```

```python
columns = []

for header in headers:
    columns.append(header.get_text(strip=True))

print(columns)
```

```python
df = pd.DataFrame(columns=columns)

print(df)
```

Table Data

```python
rows = tables[0].find_all("tr")
```

```python
for row in rows[1:]:
    cells = row.find_all("td")
    print([cell.get_text(strip=True) for cell in cells])
```

```python
from urllib.parse import urljoin

data = []

for row in rows[1:]:
    cells = row.find_all("td")
    row_data = []

    for cell in cells:
        image = cell.find("img")

        if image:
            image_url = urljoin(url, image.get("src"))
            row_data.append(image_url)
        else:
            row_data.append(cell.get_text(strip=True))

    data.append(row_data)

print(data)
```

```python
df = pd.DataFrame(data, columns=columns)
```
