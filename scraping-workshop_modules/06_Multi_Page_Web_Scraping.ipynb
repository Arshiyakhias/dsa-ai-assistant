```python
import requests
from bs4 import BeautifulSoup
import json
import os
import time
from urllib.parse import urljoin
```

```python
BOOK_URL = "https://pressbooks.palni.org/anopenguidetodatastructuresandalgorithms/"

response = requests.get(BOOK_URL)

print(response.status_code)
```

```python
soup = BeautifulSoup(response.content, "html.parser")

print(soup.title.get_text(strip=True))
```

Find all the links

```python
links = soup.find_all("a")

print("Total links:", len(links))
```

Print a few 

```python
for link in links[:10]:
    print(link.get_text(strip=True))
    print(link.get("href"))
    print("-" * 40)
```

Filter out only the chapter urls

```python
chapters = []

for link in links:
    text = link.get_text(strip=True)
    href = link.get("href")

    if text and href and "/chapter/" in href:
        chapters.append({
            "title": text,
            "url": urljoin(BOOK_URL, href)
        })
```

```python
print(f"Found {len(chapters)} chapters.")
```

Print a few 

```python
for chapter in chapters[:5]:
    print(chapter)
```

Why urljoin()?


A webpage might give you a relative url eg: /chapter/introduction/

instead of the full URL.

So:

urljoin(BOOK_URL, href)

converts it into the complete URL.

Visiting the First Chapter

```python
chapter_url = chapters[0]["url"]

response = requests.get(chapter_url)

print(response.status_code)
```

```python
chapter_soup = BeautifulSoup(response.content, "html.parser")
```

```python
print(chapter_soup.title.get_text(strip=True))
```

Find the main content

```python
print(chapter_soup.prettify()[:5000])
```

```python
main = chapter_soup.find("main")

print(main)
```

```python
content = main.get_text("\n", strip=True)

print(content[:2000])
```

Extracting one chapter completely

```python
chapter_url = chapters[0]["url"]

response = requests.get(chapter_url)

chapter_soup = BeautifulSoup(response.content, "html.parser")

main = chapter_soup.find("main")

content = main.get_text("\n", strip=True)

print(content[:2000])
```

Loop through all the chapters

```python
book_data = []

for number, chapter in enumerate(chapters, start=1):

    print(f"Scraping Chapter {number}: {chapter['title']}")

    response = requests.get(chapter["url"])

    if response.status_code == 200:

        chapter_soup = BeautifulSoup(
            response.content,
            "html.parser"
        )

        main = chapter_soup.find("main")

        if main:
            content = main.get_text("\n", strip=True)

            book_data.append({
                "chapter_number": number,
                "title": chapter["title"],
                "url": chapter["url"],
                "content": content
            })

    time.sleep(1)
```

Check the Scarped Data

```python
print(book_data[0])
```

Save as JSON

```python
import os

os.makedirs("data", exist_ok=True)

with open("data/book.json", "w", encoding="utf-8") as file:
    json.dump(
        book_data,
        file,
        indent=2,
        ensure_ascii=False
    )
```
