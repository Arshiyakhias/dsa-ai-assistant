```python
import requests
from bs4 import BeautifulSoup

url = "https://worldpopulationreview.com/country-rankings/largest-countries-in-the-world"

response = requests.get(url)

soup = BeautifulSoup(response.content, "html.parser")
```

Tag

```python
print(soup.h1)
```

```python
print(type(soup.h1))
```

Navigable String

A NavigableString represents the text inside an HTML tag 

```python
print(soup.title.string)
```

```python
print(type(soup.title.string))
```

HTML Attributes

<a href="https://example.com" class="link">Example</a>

a                → tag
href             → attribute
https://...      → attribute value
class            → attribute
link             → attribute value

```python
link = soup.a

print(link)
print(link.attrs)
```
